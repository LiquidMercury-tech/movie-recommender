```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
```

```python
data=pd.read_csv('/kaggle/input/datasets/dipson/moviers/dataset.csv')
```

```python
data.head()
```

```python
data.columns
```

```python
data.info()
```

```python
data['tags'] = data['genre'] + data['overview']
```

```python
movies=data[['id','title','genre','overview','tags']]
```

```python
movies.head()
```

```python
movies = movies.drop(columns=['genre','overview'])
```

```python
movies.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=10000, stop_words='english')
```

```python
cv
```

```python
vec=cv.fit_transform(movies['tags'].values.astype('U')).toarray()
```

```python
vec
```

```python
vec.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim=cosine_similarity(vec)
```

```python
sim
```

```python
movies[movies['title']=='The Shawshank Redemption']
```

```python
dist = sorted(list(enumerate(sim[0])),reverse=True,key= lambda vec:vec[1])
```

```python
dist
```

```python
for i in dist[0:5]:
    print(movies.iloc[i[0]].title)
```

```python
def recommend(movie):
    movie_index = movies[movies['title'] == movie].index[0]
    distances = sim[movie_index]
    similarity_list = sorted(list(enumerate(distances)), reverse=True, key=lambda vec: vec[1])
    for i in similarity_list[1:6]:
        print(movies.iloc[i[0]].title)
```

```python
recommend('Iron Man')
```
